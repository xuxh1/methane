```python
import os
import shutil

def cp_from_origin(ori_dir, root_dir, allowed_dirs, dry_run=True):
    """
    将 ori_dir 中的 landdata 和 restart 复制到 root_dir 下。
    """
    for dirname in allowed_dirs:
        src = os.path.join(ori_dir, dirname)
        dst = os.path.join(root_dir, dirname)

        if not os.path.exists(src):
            print(f"[跳过] 未找到源目录: {src}")
            continue

        # 若目标目录已存在，提示并跳过
        if os.path.exists(dst):
            print(f"[跳过] 目标目录已存在: {dst}")
            continue

        print(f"[复制] {src}  ->  {dst}")
        if not dry_run:
            shutil.copytree(src, dst)   # 等价于 cp -r


def rename_files_in_selected_subdirs(root_dir, old_str, new_str, allowed_dirs={"landdata", "restart"}, dry_run=True):
    """
    只遍历 root_dir 下的 landdata 和 restart，修改文件名：old_str → new_str
    """
    
    count = 0

    for sub in os.listdir(root_dir):
        sub_path = os.path.join(root_dir, sub)
        
        if os.path.isdir(sub_path) and sub in allowed_dirs:
            print(f"\n>>> 正在扫描目录: {sub_path}")
            
            for dirpath, dirnames, filenames in os.walk(sub_path):
                for filename in filenames:
                    if old_str in filename:
                        old_path = os.path.join(dirpath, filename)
                        new_filename = filename.replace(old_str, new_str)
                        new_path = os.path.join(dirpath, new_filename)

                        if os.path.exists(new_path):
                            print(f"[跳过] 目标文件已存在: {new_path}")
                            continue

                        if dry_run:
                            print(f"[预览] {old_path} -> {new_path}")
                        else:
                            os.rename(old_path, new_path)
                            print(f"[重命名] {old_path} -> {new_path}")

                        count += 1

    if count == 0:
        print("\n未找到需要修改的文件。")
    else:
        print(f"\n共{'预览' if dry_run else '完成'}重命名 {count} 个文件。")


if __name__ == "__main__":
    ori_directory = "/share/home/dq076/data/cases/globe/g1"
    target_directory = "/share/home/dq076/data/cases/globe/g2"

    old_pattern = "g1_"
    new_pattern = "g2_"
    dry_run_mode = False

    allowed_dirs = {"landdata", "restart"}

    print("==== 1. 复制 landdata/ 和 restart/ ====")
    # cp_from_origin(ori_directory, target_directory, allowed_dirs, dry_run=dry_run_mode)

    print("\n==== 2. 重命名 g2 中的文件 ====")
    rename_files_in_selected_subdirs(target_directory, old_pattern, new_pattern, allowed_dirs, dry_run=dry_run_mode)
```
